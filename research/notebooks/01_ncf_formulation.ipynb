# Neural Collaborative Filtering (NCF) — Mathematical Foundations

This notebook formalizes the mathematical basis of Neural Collaborative Filtering (NCF) as described in He et al. (2017).

## 1. Problem Definition

- Users: U = {1, ..., M}
- Items: I = {1, ..., N}
- Interaction Matrix: Y ∈ {0,1}^(M×N)

Implicit Feedback:
- Y[u,i] = 1 → interaction exists
- Y[u,i] = 0 → unobserved (NOT negative)

Sparsity: ~95% missing values

## 2. Model Objective

Learn a function:

ŷ[u,i] = f(u, i | Θ)

Where:
- ŷ[u,i] ∈ [0,1]
- Θ = model parameters
- Output = probability of interaction

## 3. Embedding Representation

- User embedding: p_u ∈ R^k
- Item embedding: q_i ∈ R^k

Embedding matrices:
- P ∈ R^(M×k)
- Q ∈ R^(N×k)

## 4. Forward Pass (NCF Architecture)

### Step 1: Embedding Lookup
p_u = P[u,:]
q_i = Q[i,:]

### Step 2: Concatenation
z₀ = [p_u ⊕ q_i] ∈ R^(2k)

### Step 3: MLP Layers
z₁ = ReLU(W₁ z₀ + b₁)
...
z_L = ReLU(W_L z_{L-1} + b_L)

### Step 4: Output Layer
ŷ = σ(wᵀ z_L + b)

## 5. Loss Function — Binary Cross Entropy

L(y, ŷ) = -[y log(ŷ) + (1-y) log(1-ŷ)]

Total Loss:

L_total = -Σ log(ŷ_pos) - Σ log(1 - ŷ_neg)

Negative Sampling:
- 4 negatives per positive

## 6. Key Mathematical Concepts

### Chain Rule
∂L/∂W = (∂L/∂z) · (∂z/∂W)

### Sigmoid Function
σ(x) = 1 / (1 + e^{-x})

Derivative:
σ'(x) = σ(x)(1 - σ(x))

### ReLU Function
ReLU(x) = max(0, x)

Derivative:
1 if x > 0 else 0

### Gradient (Linear Layer)
z = Wx + b

∂L/∂W = δ xᵀ
∂L/∂b = δ
∂L/∂x = Wᵀ δ

  - 99 sampled negatives
