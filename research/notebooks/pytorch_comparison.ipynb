```python
import sys
sys.path.insert(0, '/Users/rahulkpkurup/Learning/ML-Git-Portfolio/ncf-recommender-system-study')

import numpy as np
import torch
from models.scratch_ncf.forward import Scratch_NCF
from models.scratch_ncf.loss import BCELoss
from models.pytorch_ncf.model import PyTorchNCF, get_device
```

```python
# Initialize Scratch NCF Model
n_users = 100
n_items = 500
emb_dim = 32
hidden = [64, 32]

scratch_model = Scratch_NCF(
    n_users=n_users,
    n_items=n_items,
    emb_dim=emb_dim,
    hidden=hidden,
    seed=42
)

loss_fn = BCELoss()

print("Scratch NCF Model initialized:")
print(f"  Users: {n_users}, Items: {n_items}, Embedding dim: {emb_dim}")
print(f"  Hidden layers: {hidden}")
```

```python
# Initialize PyTorch NCF Model
device = get_device()

pt_model = PyTorchNCF(
    n_users=n_users,
    n_items=n_items,
    emb_dim=emb_dim,
    hidden=hidden,
    dropout=0.0  # No dropout for fair comparison
)

# Initialize embeddings by doing a dummy forward pass
with torch.no_grad():
    dummy_u = torch.tensor([0], dtype=torch.long)
    dummy_i = torch.tensor([0], dtype=torch.long)
    _ = pt_model(dummy_u, dummy_i)

# Then move to device
pt_model = pt_model.to(device)
pt_model.eval()  # Set to eval mode for inference

print(f"PyTorch NCF Model initialized on device: {device}")
print(f"  Users: {n_users}, Items: {n_items}, Embedding dim: {emb_dim}")
print(f"  Hidden layers: {hidden}")
```

```python
# Synchronize weights: copy scratch_model weights → pt_model
with torch.no_grad():
    # Copy embedding weights
    pt_model.user_emb.weight.copy_(torch.from_numpy(scratch_model.user_emb.W).float())
    pt_model.item_emb.weight.copy_(torch.from_numpy(scratch_model.item_emb.W).float())
    
    # Extract Linear layers from mlp Sequential (they are at indices 0, 2, 4, ...)
    mlp_linear_layers = [module for module in pt_model.mlp.modules() if isinstance(module, torch.nn.Linear)]
    
    # Copy weights for hidden layers from scratch_model.linears[0:2]
    for pt_layer, scratch_layer in zip(mlp_linear_layers, scratch_model.linears[:-1]):
        pt_layer.weight.copy_(torch.from_numpy(scratch_layer.W).float())
        pt_layer.bias.copy_(torch.from_numpy(scratch_layer.b).float())
    
    # Copy weights for output layer
    pt_model.output.weight.copy_(torch.from_numpy(scratch_model.linears[-1].W).float())
    pt_model.output.bias.copy_(torch.from_numpy(scratch_model.linears[-1].b).float())

print("✓ Weights synchronized: PyTorch model now uses scratch model's weights")
print("✓ Logit outputs should now be numerically identical (within floating-point precision)")
```

```python
# Verify numerical parity — compare loss on a single batch
import numpy as np
import torch

# Same input
uid, iid, y = 42, 100, 1.0

# Scratch
scratch_logit = scratch_model.forward(uid, iid)
scratch_loss  = loss_fn.forward(scratch_logit, y)

# PyTorch
u_t = torch.tensor([uid], dtype=torch.long).to(device)
i_t = torch.tensor([iid], dtype=torch.long).to(device)
with torch.no_grad():
    pt_logit = pt_model(u_t, i_t).item()
pt_loss  = torch.nn.functional.binary_cross_entropy_with_logits(
    torch.tensor([pt_logit]), torch.tensor([y])
).item()

print(f"Scratch logit: {scratch_logit:.6f}   loss: {scratch_loss:.6f}")
print(f"PyTorch logit: {pt_logit:.6f}   loss: {pt_loss:.6f}")
print(f"Logit diff: {abs(scratch_logit - pt_logit):.2e}")
```

```python
import json
import matplotlib.pyplot as plt
import matplotlib.style as style
from pathlib import Path

# ------------------------------------------------------------------
# Cell 1 — Load both logs with dynamic path resolution
# ------------------------------------------------------------------
ROOT_DIR = Path.cwd()
if ROOT_DIR.name != "ncf-recommender-system-study" and (ROOT_DIR.parent / "evaluation").exists():
    ROOT_DIR = ROOT_DIR.parent

eval_dir = ROOT_DIR / "evaluation"

with open(eval_dir / "scratch_ncf_log.json", "r") as f:
    scratch = json.load(f)

with open(eval_dir / "pytorch_ncf_log.json", "r") as f:
    pytorch = json.load(f)

# Extract loss data
s_epochs = [e["epoch"] for e in scratch["epochs"]]
s_loss   = [e["loss"]  for e in scratch["epochs"]]

p_epochs = [e["epoch"] for e in pytorch["epochs"]]
p_loss   = [e["loss"]  for e in pytorch["epochs"]]

# Extract Hit@10 data AND corresponding epoch arrays separately for each model
s_hit_epochs = [e["epoch"] for e in scratch["epochs"] if e.get("hit_at_10") is not None]
s_hit        = [e["hit_at_10"] for e in scratch["epochs"] if e.get("hit_at_10") is not None]

p_hit_epochs = [e["epoch"] for e in pytorch["epochs"] if e.get("hit_at_10") is not None]
p_hit        = [e["hit_at_10"] for e in pytorch["epochs"] if e.get("hit_at_10") is not None]


# ------------------------------------------------------------------
# Cell 2 — Loss curve & Hit@10 curve (Figure 4.1 in your thesis)
# ------------------------------------------------------------------
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Subplot 1: Training Loss
axes[0].plot(s_epochs, s_loss, label="Scratch NCF", color="#D85A30", lw=2)
axes[0].plot(p_epochs, p_loss, label="PyTorch NCF", color="#1D9E75", lw=2)
axes[0].set_xlabel("Epoch")
axes[0].set_ylabel("BCE Loss")
axes[0].set_title("Training Loss Convergence")
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Subplot 2: Hit@10 (Paired with model-specific epoch arrays)
axes[1].plot(s_hit_epochs, s_hit, label="Scratch NCF", color="#D85A30",
             lw=2, marker="o", ms=4)
axes[1].plot(p_hit_epochs, p_hit, label="PyTorch NCF", color="#1D9E75",
             lw=2, marker="o", ms=4)
axes[1].set_xlabel("Epoch")
axes[1].set_ylabel("Hit@10")
axes[1].set_title("Recommendation Quality (Hit@10)")
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig(eval_dir / "figure_4_1_convergence.png", dpi=150, bbox_inches="tight")
plt.show()


# ------------------------------------------------------------------
# Cell 3 — Training time bar chart (Figure 4.2)
# ------------------------------------------------------------------
s_times = [e["time_sec"] for e in scratch["epochs"]]
p_times = [e["time_sec"] for e in pytorch["epochs"]]

s_total_time = sum(s_times)
p_total_time = sum(p_times)

fig, ax = plt.subplots(figsize=(6, 4))
bars = ax.bar(["Scratch NCF\n(NumPy)", "PyTorch NCF\n(MPS)"],
              [s_total_time, p_total_time],
              color=["#D85A30", "#1D9E75"], width=0.5)

ax.set_ylabel("Total training time (seconds)")
ax.set_title(f"Training Efficiency: {len(s_epochs)} Epochs")

# Annotate total seconds above each bar safely dynamically offset
y_max = max(s_total_time, p_total_time)
for i, v in enumerate([s_total_time, p_total_time]):
    ax.text(i, v + (y_max * 0.02), f"{v:.1f}s", ha="center", fontsize=11, fontweight="bold")

# Add overhead headroom for text labels
ax.set_ylim(0, y_max * 1.15)

plt.tight_layout()
plt.savefig(eval_dir / "figure_4_2_training_time.png", dpi=150, bbox_inches="tight")
plt.show()
```
