# Libraries

```python
import sys
import os
import torch
import numpy as np
```

# Path setup

```python
sys.path.append(os.path.abspath(os.path.join('..')))
```

# Imports

```python
from models.scratch_ncf.forward import ScratchNCF
from models.scratch_ncf.loss import BCELoss 
from models.pytorch_ncf.model import PyTorchNCF
```

# Parameters

```python
num_users = 6040 
num_items = 3952
```

# Initialization

```python
scratch_model = ScratchNCF(num_users, num_items)
loss_fn = BCELoss()
```

```python

pt_model = PyTorchNCF(num_users, num_items)
pt_model.eval()
```

# Both models and loss function initialized successfully

# Parity Test

## Same input

```python
uid, iid, y = 42, 100, 1.0
```

# Scratch Execution

```python
scratch_logit = scratch_model.forward(uid, iid)
scratch_loss  = loss_fn.forward(scratch_logit, y)
```

# PyTorch Execution

```python
u_t = torch.tensor([uid], dtype=torch.long)
i_t = torch.tensor([iid], dtype=torch.long)
```

# Calling the model directly (assuming its forward takes (u, i))

```python
with torch.no_grad():
    pt_output = pt_model(u_t, i_t)
    pt_logit = pt_output.item()
```

# Use PyTorch's built-in BCE with Logits for comparison

```python
pt_loss = torch.nn.functional.binary_cross_entropy_with_logits(
    torch.tensor([pt_logit]), 
    torch.tensor([y])
).item()
```

```python
print(f"Scratch logit: {scratch_logit:.6f}   loss: {scratch_loss:.6f}")
print(f"PyTorch logit: {pt_logit:.6f}   loss: {pt_loss:.6f}")
print(f"---")
print(f"Logit diff: {abs(scratch_logit - pt_logit):.2e}")
```
