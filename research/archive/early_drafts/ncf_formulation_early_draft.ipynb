# NCF Mathematical Formulation

## Problem Definition

- **Users:** $U = \{1, ..., M\}$, $M = 6040$ (MovieLens 1M)
- **Items:** $I = \{1, ..., N\}$, $N = 3706$
- **Interaction matrix:** $Y \in \{0,1\}^{M \times N}$
- **Sparsity:** ~95% unobserved

## Hypothesis

A NumPy NCF and PyTorch NCF trained on identical data with identical hyperparameters will converge to equivalent Hit@10 within ±2%, but PyTorch will train 10-50x faster due to vectorised autograd.

## Model: $f(u, i | \Theta) = \sigma(\text{MLP}([p_u \oplus q_i]))$

- $p_u = P[u,:]$, $P \in \mathbb{R}^{M \times k}$ — user embedding
- $q_i = Q[i,:]$, $Q \in \mathbb{R}^{N \times k}$ — item embedding
- **Concat:** $z_0 = [p_u; q_i] \in \mathbb{R}^{2k}$
- **Hidden:** $z_l = \text{ReLU}(W_l z_{l-1} + b_l)$
- **Output:** $\hat{y} = \sigma(w^T z_L + b)$

## Loss: Binary Cross Entropy with Negative Sampling

$$L = -\sum_{(u,i) \in Y^+} \log(\hat{y}_{ui}) - \sum_{(u,j) \in Y^-} \log(1 - \hat{y}_{uj})$$

**Negative sampling ratio:** 4 negatives per positive

## Evaluation Protocol

- **Leave-one-out:** hold last interaction per user as test
- **Sample:** 99 random negatives per test interaction
- **Rank:** all 100, compute: Hit@10, NDCG@10
